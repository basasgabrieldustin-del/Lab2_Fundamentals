```python
# TASK 1: REDACTION PROTOCOL
target_name = "BASAS"       # Your SURNAME in UPPERCASE
SEED_NUM = 2                 # Last digit of STUDENT_ID
redacted_name = ""

for char in target_name:
    if char in "AEIOU":
        redacted_name += str(SEED_NUM)
    else:
        redacted_name += char

print("Original Surname:", target_name)
print("SEED_NUM Used:", SEED_NUM)
print("Redacted Output:", redacted_name)
```

```python
# TASK 2: FORENSIC SIGNATURE
first_letter = "B"
SEED_NUM = 2
loops = 5

ASCII = ord(first_letter)
verification_hash = (ASCII * SEED_NUM) - loops

print("First Letter ASCII:", ASCII)
print("SEED_NUM:", SEED_NUM)
print("Total Loops:", loops)
print("Verification Hash:", verification_hash)
```

```python
# TASK 3: SECURE INPUT LOOP
correct_password = "ADMIN"
attempt_count = 0
execution_log = ""

while True:
    attempt_count += 1
    user_input = input("Enter password: ")
    
    if user_input == correct_password:
        execution_log += f"Attempt {attempt_count}: '{user_input}' → SUCCESS"
        print("Unlocked.")
        break
    else:
        execution_log += f"Attempt {attempt_count}: '{user_input}' → Access Denied\n"
        print("Access Denied.")

print("\n--- ASSESSMENT DATA ---")
print("Correct Password Set:", correct_password)
print("Input Attempts:", attempt_count)
print("Execution Log:\n", execution_log)
print("Attempts Before Success:", attempt_count)
```
